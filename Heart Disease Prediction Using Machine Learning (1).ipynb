```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size =0.2,random_state = 42)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaler
```

```python
x_train = scaler.fit_transform(x_train)
```

```python
x_test = scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
regression = LogisticRegression()
regression
```

```python
regression.fit(x_train,y_train)
```

```python
lr_pred= regression.predict(x_test)
```

```python
lr_pred
```

```python
from sklearn.ensemble  import RandomForestClassifier
```

```python
rf = RandomForestClassifier(n_estimators=100,random_state=42)
rf
```

```python
rf.fit(x_train,y_train)
```

```python
rf_pred = rf.predict(x_test)
rf_pred
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt = DecisionTreeClassifier()
dt
```

```python
dt.fit(x_train,y_train)
```

```python
dt_pred = dt.predict(x_test)
dt_pred
```

```python
from sklearn.ensemble import GradientBoostingClassifier
gr = GradientBoostingClassifier()
gr
```

```python
gr.fit(x_train,y_train)
```

```python
gr_pred = gr.predict(x_test)
gr_pred
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score
```

```python
print("logistics_regression:",accuracy_score(y_test,lr_pred))
print("random_forest:",accuracy_score(y_test,rf_pred))
print("decision_tree:",accuracy_score(y_test,dt_pred))
print("gradientboost:",accuracy_score(y_test,gr_pred))
```

```python
print("Logistic Regression :", recall_score(y_test, lr_pred))
print("Random Forest :", recall_score(y_test, rf_pred))
print("Decision Tree :", recall_score(y_test, dt_pred))
print("Gradient Boosting :", recall_score(y_test, gr_pred))
```

```python
from sklearn.metrics import precision_score
```

```python
print("Logistic Regression :", precision_score(y_test, lr_pred))
print("Random Forest :", precision_score(y_test, rf_pred))
print("Decision Tree :", precision_score(y_test, dt_pred))
print("Gradient Boosting :", precision_score(y_test, gr_pred))
```

```python
from sklearn.metrics import f1_score
```

```python
print("Logistic Regression :", f1_score(y_test, lr_pred))
print("Random Forest :", f1_score(y_test, rf_pred))
print("Decision Tree :", f1_score(y_test, dt_pred))
print("Gradient Boosting :", f1_score(y_test, gr_pred))
```

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion = confusion_matrix(y_test,rf_pred)
confusion
```

```python
sns.heatmap(confusion, annot=True, fmt="d")

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Random Forest Confusion Matrix")

plt.show()
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "Decision Tree",
        "Gradient Boosting"
    ],

    "Accuracy": [
        accuracy_score(y_test, lr_pred),
        accuracy_score(y_test, rf_pred),
        accuracy_score(y_test, dt_pred),
        accuracy_score(y_test, gr_pred)
    ],

    "Precision": [
        precision_score(y_test, lr_pred),
        precision_score(y_test, rf_pred),
        precision_score(y_test, dt_pred),
        precision_score(y_test, gr_pred)
    ],

    "Recall": [
        recall_score(y_test, lr_pred),
        recall_score(y_test, rf_pred),
        recall_score(y_test, dt_pred),
        recall_score(y_test, gr_pred)
    ],

    "F1": [
        f1_score(y_test, lr_pred),
        f1_score(y_test, rf_pred),
        f1_score(y_test, dt_pred),
        f1_score(y_test, gr_pred)
    ]
})

print(results)
```
